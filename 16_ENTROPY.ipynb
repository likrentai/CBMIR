# 16 ENTROPY: randomness and entropy of the fault process

Two questions are kept separate throughout. **Distributional entropy** asks how concentrated fault episodes are across messages. **Temporal entropy** asks how regular the daily count sequence is over time.

Days on which the locomotive was out of service for annual maintenance are masked as missing, and rolling windows never span that period. A final section repeats the measures without the blower supervision message.

Input `build/fault_episodes_v4.csv`; outputs in `OUTPUT/v4/`. Requires `antropy` and `statsmodels`.

Paper: Sections 2.8 and 3.4; Tables 12 to 14; Figures 9 to 11.

```python
import os, warnings, numpy as np, pandas as pd
warnings.filterwarnings("ignore")
import antropy as ant
from scipy.stats import entropy, norm, spearmanr, pearsonr
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.tsa.stattools import grangercausalitytests
import statsmodels.api as sm

CFG = dict(
    episodes="build/fault_episodes_v4.csv",
    outdir="OUTPUT/v4",
    gap_min_days=7,
    window=30,        # rolling window, days
    m=2,              # embedding dimension for ApEn / SampEn
    r_mult=0.2,       # tolerance as a multiple of the window SD
    order=3,          # permutation entropy order
    delay=1,          # permutation entropy delay
)

os.makedirs(CFG["outdir"], exist_ok=True)
```

## 1. Daily counts and in-service segments


```python
def load(cfg):
    ep = pd.read_csv(cfg["episodes"], parse_dates=["onset"])
    s = (ep.set_index("onset").resample("D").size()
           .asfreq("D").fillna(0).astype(float))
    s.name = "faults"
    zero = s.eq(0)
    block = (zero != zero.shift()).cumsum()
    masked = []
    for _, idx in s.groupby(block).groups.items():
        idx = pd.DatetimeIndex(idx)
        if zero.loc[idx].all() and len(idx) >= cfg["gap_min_days"]:
            s.loc[idx] = np.nan
            masked.append((idx.min().date(), idx.max().date(), len(idx)))
    print(f"{len(s)} days {s.index.min().date()} to {s.index.max().date()};"
          f" observed {int(s.notna().sum())}, masked {int(s.isna().sum())}")
    for a, b, n in masked:
        print(f"  out of service {a} to {b} ({n} days)")
    return ep, s
```

```python
def segments(s, minlen):
    """Contiguous runs of observed days, as a list of Series."""
    ok = s.notna()
    blk = (ok != ok.shift()).cumsum()
    out = []
    for _, g in s.groupby(blk):
        if g.notna().all() and len(g) >= minlen:
            out.append(g)
    return out
```

```python
ep, s = load(CFG)
segs = segments(s, CFG['window'])
print('in-service segments:', [(g.index.min().date(), g.index.max().date(), len(g)) for g in segs])
```

## 2. Randomness of the daily count sequence

The Wald-Wolfowitz runs test asks whether values alternate around the median
more or less often than chance would produce. The Ljung-Box test asks whether
the series is serially correlated. They answer different questions, so both are
reported, together with the short-lag autocorrelations that drive them.


```python
def runs_test(x):
    x = np.asarray(x, float)
    med = np.median(x)
    b = (x >= med).astype(int)
    n1, n2 = int(b.sum()), int(len(b) - b.sum())
    runs = 1 + int(np.sum(np.abs(np.diff(b))))
    mu = (2 * n1 * n2) / (n1 + n2) + 1
    sd = np.sqrt((2 * n1 * n2 * (2 * n1 * n2 - n1 - n2)) /
                 (((n1 + n2) ** 2) * (n1 + n2 - 1)))
    z = (runs - mu) / sd
    p = 2 * (1 - norm.cdf(abs(z)))
    return dict(n=len(x), runs=runs, expected=mu, z=z, p=p)
```

```python
def randomness(s, cfg):
    obs = s.dropna()
    print("\nRANDOMNESS OF THE DAILY COUNT SEQUENCE")
    r = runs_test(obs.values)
    print(f"  Wald-Wolfowitz runs test  n={r['n']}  runs={r['runs']}"
          f"  expected={r['expected']:.2f}  z={r['z']:.3f}  p={r['p']:.4f}")
    from statsmodels.tsa.stattools import acf as _acf
    r_ = _acf(s, nlags=7, fft=False, missing="conservative")
    print("  gap-aware ACF  " + "  ".join(f"lag{k}={r_[k]:+.3f}" for k in range(1, 8)))
    lb = acorr_ljungbox(obs, lags=[7, 14, 30], return_df=True)
    for lag, row in lb.iterrows():
        print(f"  Ljung-Box lag {lag:>2}       Q={row.lb_stat:8.3f}  p={row.lb_pvalue:.4f}")
    return r, lb
```

```python
runs, lb = randomness(s, CFG)
```

## 3. Distributional entropy

Shannon entropy over the frequency of each fault message, reported against its maximum and in normalised form. The category-level values of Table 14 are computed by `TABLE1_CATEGORIES.py` over the seven categories of Table 1; the category grouping printed in this section is an earlier, six-category version and is superseded by that script.

```python
def distributional_entropy(ep, col, label):
    counts = ep[col].value_counts()
    p = counts.values / counts.values.sum()
    h = entropy(p, base=2)
    hmax = np.log2(len(p))
    print(f"  {label:<28s} H={h:.3f} bits   Hmax={hmax:.3f}"
          f"   normalised={h/hmax:.3f}   categories={len(p)}")
    return dict(kind=label, H=h, Hmax=hmax, H_norm=h / hmax, k=len(p))
```

```python
CATEGORY = {0: "Ventilation", 2: "Protection system", 3: "Protection system", 5: "Traction and power conversion",
            6: "Traction and power conversion", 7: "Traction and power conversion", 9: "Braking and pneumatic",
            10: "Protection system", 11: "Protection system", 12: "Traction and power conversion",
            13: "Protection system", 14: "Traction and power conversion", 15: "Auxiliary",
            17: "Power supply quality", 19: "Power supply quality", 16: "Auxiliary", 8: "Protection system"}
ep['category'] = ep['cluster'].map(CATEGORY)
assert ep['category'].notna().all(), 'a cluster has no Table 1 category'
nb_ = ep[ep['fault'] != 'AUX CAP MACH RM BLOW1 NOT OFF']
dist = [distributional_entropy(ep, 'fault', 'Fault label'),
        distributional_entropy(ep, 'category', 'Fault category (Table 1)'),
        distributional_entropy(nb_, 'fault', 'Fault label, without blower'),
        distributional_entropy(nb_, 'category', 'Fault category, without blower')]
pd.DataFrame(dist).to_csv(f"{CFG['outdir']}/distributional_entropy.csv", index=False)
```

## 4. Temporal entropy of the daily count series

Approximate, sample and permutation entropy, computed per in-service segment so
that no embedding vector spans the maintenance period. Parameters are fixed in
`CFG`: embedding dimension m, tolerance as a multiple of the series standard
deviation, and permutation order and delay.


```python
def temporal_entropy(x, cfg):
    x = np.asarray(x, float)
    return dict(
        approximate=ant.app_entropy(x, order=cfg["m"]),
        sample=ant.sample_entropy(x, order=cfg["m"]),
        permutation=ant.perm_entropy(x, order=cfg["order"], delay=cfg["delay"], normalize=True),
    )
```

```python
rows = []
for g in segs:
    te = temporal_entropy(g.values, CFG)
    te.update(start=g.index.min().date(), end=g.index.max().date(), days=len(g))
    rows.append(te)
temporal = pd.DataFrame(rows)[['start', 'end', 'days', 'approximate', 'sample', 'permutation']]
temporal.to_csv(f"{CFG['outdir']}/temporal_entropy.csv", index=False)
print(temporal.round(3).to_string(index=False))
```

## 5. Rolling-window entropy

Windows are formed inside a single in-service segment only. Windows that would
straddle the maintenance period are not formed at all.


```python
def rolling_entropy(s, cfg):
    """Rolling windows that never span an out-of-service period."""
    w = cfg["window"]
    rows = []
    for si, seg in enumerate(segments(s, w)):
        v = seg.values
        for i in range(len(v) - w + 1):
            win = v[i:i + w]
            te = temporal_entropy(win, cfg)
            rows.append(dict(date=seg.index[i + w - 1], segment=si,
                             rate=win.mean(),
                             approximate=te["approximate"],
                             sample=te["sample"],
                             permutation=te["permutation"]))
    return pd.DataFrame(rows).set_index("date")
```

```python
roll = rolling_entropy(s, CFG)
roll.to_csv(f"{CFG['outdir']}/rolling_entropy.csv")
print(f'{len(roll)} windows across {roll.segment.nunique()} segment(s)')
print(roll.groupby('segment').agg(n=('rate','size'), first=('rate','count')).to_string())
```

## 6. Fault rate against entropy

Successive windows share 29 of their 30 days, so ordinary least squares standard
errors are far too small. A HAC correction with a lag truncation of one window
length is reported alongside them, and the Granger test is run within the longest
segment so that no lag crosses the maintenance period.


```python
def rate_vs_entropy(roll, measure="permutation", cfg=CFG, seg_id=None):
    d = roll[["rate", measure]].dropna()
    if seg_id is not None:
        d = d[roll.loc[d.index, "segment"] == seg_id]
    pr, pp = pearsonr(d["rate"], d[measure])
    sr, sp = spearmanr(d["rate"], d[measure])
    X = sm.add_constant(d[["rate"]])
    ols = sm.OLS(d[measure], X).fit()
    # windows overlap by window-1 days, so OLS standard errors are far too small; HAC correction with a lag truncation of one window length
    hac = ols.get_robustcov_results(cov_type="HAC", maxlags=cfg["window"])
    try:
        g = grangercausalitytests(d[[measure, "rate"]].values, maxlag=7)
        gp = min(v[0]["ssr_ftest"][1] for v in g.values())
    except Exception as e:
        print("   granger failed:", type(e).__name__, e)
        gp = np.nan
    print(f"\nFAULT RATE vs {measure.upper()} ENTROPY  (n={len(d)})")
    print(f"  Pearson  r={pr:.3f}  p={pp:.4f}")
    print(f"  Spearman r={sr:.3f}  p={sp:.4f}")
    print(f"  OLS      R2={ols.rsquared:.4f}  slope p={ols.pvalues['rate']:.4f}"
          f"   HAC slope p={hac.pvalues[1]:.4f}")
    print(f"  Granger  best p over lags 1-7 = {gp:.4f}")
    return dict(measure=measure, n=len(d), pearson_r=pr, pearson_p=pp,
                spearman_r=sr, spearman_p=sp, r2=ols.rsquared,
                slope_p=float(ols.pvalues["rate"]),
                slope_p_hac=float(hac.pvalues[1]), granger_p=gp)
```

```python
stats = [rate_vs_entropy(roll, 'permutation', CFG, seg_id=0),
         rate_vs_entropy(roll, 'approximate', CFG, seg_id=0)]
pd.DataFrame(stats).to_csv(f"{CFG['outdir']}/rate_vs_entropy.csv", index=False)
```

## Without the blower supervision message

The same randomness and temporal-entropy measures on the daily count of the other messages, with the out-of-service days of the full series masked.

```python
E = dict(episodes='build/fault_episodes_v4.csv', outdir='OUTPUT/v4', gap_min_days=7, window=30, m=2, r_mult=0.2, order=3, delay=1)   # settings of the first cell (the figure cells redefine the configuration)
ep_all = pd.read_csv(E['episodes'], parse_dates=['onset'])
ep_nb = ep_all[ep_all['fault'] != 'AUX CAP MACH RM BLOW1 NOT OFF']
s_full = load(E)[1]
s_nb = (ep_nb.set_index('onset').resample('D').size().reindex(s_full.index).fillna(0).astype(float))
s_nb[s_full.isna()] = np.nan
print(f'other labels: {int(s_nb.notna().sum())} observed days, mean {s_nb.mean():.2f} per day')
runs_nb, lb_nb = randomness(s_nb, E)
rows = []
for g in segments(s_nb, E['window']):
    te = temporal_entropy(g.values, E)
    te.update(start=g.index.min().date(), end=g.index.max().date(), days=len(g)); rows.append(te)
t_nb = pd.DataFrame(rows)[['start', 'end', 'days', 'approximate', 'sample', 'permutation']]
print(t_nb.round(3).to_string(index=False))
t_nb.to_csv(f"{E['outdir']}/temporal_entropy_noblower.csv", index=False)
pd.DataFrame([dict(runs_test, **{'lb7_p': lb_nb.loc[7, 'lb_pvalue'], 'lb14_p': lb_nb.loc[14, 'lb_pvalue']}) for runs_test in [runs_nb]]).to_csv(f"{E['outdir']}/randomness_noblower.csv", index=False)
```
